```python
import pandas as pd
import matplotlib.pyplot as plt

audio_df = pd.read_csv("experiments/logs/audio_only_training.csv")
vibration_df = pd.read_csv("experiments/logs/vibration_training.csv")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(audio_df['val_si_sdr'], label='audio', linewidth=2)
plt.plot(vibration_df['val_si_sdr'], label='vibration', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Validation SI-SDR (dB)')
plt.title('Audio-Only vs Vibration-Guided Model Comparison')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(audio_df['train_si_sdr'], label='audio', linewidth=2)
plt.plot(vibration_df['train_si_sdr'], label='vibration', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Train SI-SDR (dB)')
plt.title('Audio-Only vs Vibration-Guided Model Comparison')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
